```
public interface IOrganizable
{
    string OrganizerName { get; set; }
    decimal Budget { get; set; }
    void DisplayBudgetInfo();
}

public interface ILivestreamable
{
    string StreamUrl { get; set; }
    void StartStream();
    void EndStream();
}

public class Event
{
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }
    public int Capacity { get; set; } 
    public decimal TicketPrice { get; set; } 
    public bool IsCancelled { get; set; }  

    public List<string> Participants = new List<string>();

    public Event(int eventId, string eventName, DateTime date, int capacity = 100)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
        Capacity = capacity;
    }

    public virtual void RegisterParticipant(string participantName)
    {
        if (Participants.Count < Capacity)
        {
            Participants.Add(participantName);
            Console.WriteLine("Участник " + participantName + " зарегистрирован на " + EventName);
        }
        else
        {
            Console.WriteLine("Нет свободных мест на " + EventName);
        }
    }


    public void RegisterParticipant(string participantName, string vipCode)
    {
        if (Participants.Count < Capacity)
        {
            Participants.Add(participantName + " [VIP]");
            Console.WriteLine("VIP-участник " + participantName + " зарегистрирован по коду " + vipCode);
        }
    }

    public virtual void CancelParticipant(string participantName)
    {
        if (Participants.Remove(participantName))
        {
            Console.WriteLine("Регистрация участника " + participantName + " отменена.");
        }
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine("Отправка уведомления для " + Participants.Count + " участников мероприятия " + EventName);
    }


    public void NotifyParticipants(string customMessage)
    {
        Console.WriteLine("Уведомление для " + EventName + ": " + customMessage);
    }

    public virtual string GetEventDetails()
    {
        return "ID: " + EventId + ", Название: " + EventName + ", Дата: " + Date.ToShortDateString() + ", Вместимость: " + Capacity;
    }

    public void CancelEvent()
    {
        IsCancelled = true;
        Console.WriteLine("Мероприятие " + EventName + " отменено.");
    }
}

public class Conference : Event
{
    public int SpeakersCount { get; set; }
    public string HallNumber { get; set; }

    public Conference(int eventId, string eventName, DateTime date, int speakersCount, string hallNumber = "Главный зал")
        : base(eventId, eventName, date)
    {
        SpeakersCount = speakersCount;
        HallNumber = hallNumber;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine("Отправка программы конференции " + EventName + " с участием " + SpeakersCount + " спикеров в зале " + HallNumber + ".");
    }

    public void AddSpeaker()
    {
        SpeakersCount++;
        Console.WriteLine("Добавлен новый спикер! Всего спикеров: " + SpeakersCount);
    }
}

public class Festival : Event, IOrganizable
{
    public int BandsCount { get; set; }
    public string OrganizerName { get; set; } 
    public decimal Budget { get; set; } 

    public Festival(int eventId, string eventName, DateTime date, int bandsCount, string organizerName, decimal budget)
        : base(eventId, eventName, date)
    {
        BandsCount = bandsCount;
        OrganizerName = organizerName;
        Budget = budget;
    }

    public void RegisterParticipant(string participantName, string genre)
    {
        Participants.Add(participantName + " (" + genre + ")");
        Console.WriteLine("Участник " + participantName + " зарегистрирован на фестиваль " + EventName + " с жанром " + genre);
    }

    public void DisplayBudgetInfo()
    {
        Console.WriteLine("Организатор фестиваля '" + EventName + "': " + OrganizerName + ", Бюджет: " + Budget + " руб.");
    }
}

public class Seminar : Event
{
    public string Topics { get; set; }
    public string SpeakerName { get; set; } 

    public Seminar(int eventId, string eventName, DateTime date, string topics, string speakerName)
        : base(eventId, eventName, date)
    {
        Topics = topics;
        SpeakerName = speakerName;
    }

    public override string GetEventDetails()
    {
        return base.GetEventDetails() + ", Темы семинара: " + Topics + ", Ведущий: " + SpeakerName;
    }
}

public class OnlineConference : Conference, ILivestreamable
{
    public string StreamUrl { get; set; }

    public OnlineConference(int eventId, string eventName, DateTime date, int speakersCount, string hallNumber, string streamUrl)
        : base(eventId, eventName, date, speakersCount, hallNumber)
    {
        StreamUrl = streamUrl;
    }

    public void StartStream()
    {
        Console.WriteLine("Онлайн-трансляция конференции " + EventName + " запущена! Ссылка: " + StreamUrl);
    }

    public void EndStream()
    {
        Console.WriteLine("Онлайн-трансляция " + EventName + " завершена.");
    }


    public override string GetEventDetails()
    {
        return base.GetEventDetails() + ", Ссылка на стрим: " + StreamUrl;
    }
}


public class EventContainer<T> where T : Event
{
    private List<T> events = new List<T>();

    public void Add(T item)
    {
        events.Add(item);
    }

    public void PrintDetails()
    {
        foreach (var item in events)
        {
            Console.WriteLine(item.GetEventDetails());
        }
    }
}


        Conference conf = new Conference(1, "Tech Conference", new DateTime(2026, 10, 15), 10, "Зал А");
        Festival fest = new Festival(2, "Rock Fest", new DateTime(2026, 7, 20), 5, "Global Events", 300000m);
        Seminar sem = new Seminar(3, "Программирование C#", new DateTime(2026, 11, 5), "ООП, Классы, Наследование", "Проф. Иванов");
        OnlineConference onlineConf = new OnlineConference(4, "C# Online Summit", new DateTime(2026, 12, 1), 3, "Веб-комната", "https://twitch.tv/event");

        Console.WriteLine("--- Регистрация ---");
        conf.RegisterParticipant("Иван");
        fest.RegisterParticipant("Анна", "Рок");
        sem.RegisterParticipant("Петр");
        onlineConf.RegisterParticipant("Олег");

        Console.WriteLine("\n--- Уведомления ---");
        conf.NotifyParticipants();
        fest.NotifyParticipants();

        Console.WriteLine("\n--- Взаимодействие с интерфейсами и доп. методами ---");
        fest.DisplayBudgetInfo();
        onlineConf.StartStream();
        onlineConf.EndStream();

        Console.WriteLine("\n--- Детали мероприятий ---");
        Console.WriteLine(conf.GetEventDetails());
        Console.WriteLine(fest.GetEventDetails());
        Console.WriteLine(sem.GetEventDetails());
        Console.WriteLine(onlineConf.GetEventDetails());

        Console.WriteLine("\n--- Проверка перегрузки методов ---");
        conf.RegisterParticipant("Сергей", "VIP-777");
        conf.NotifyParticipants("Срочное сообщение!");

        Console.WriteLine("\n--- Проверка Generic-класса ---");
        EventContainer<Event> container = new EventContainer<Event>();
        container.Add(conf);
        container.Add(fest);
        container.PrintDetails();
```
