Создать базовый класс Event в C#, который будет представлять организованные 
мероприятия. На основе этого класса разработать 2-3 производных класса, 
демонстрирующих принципы наследования и полиморфизма. В каждом из классов 
должны быть реализованы новые атрибуты и методы, а также переопределены 
некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;
    public class Event
    {
        public int EventId { get; set; }
        public string EventName { get; set; }
        public DateTime Date { get; set; }

        public List<string> Participants = new List<string>();

        public Event(int eventId, string eventName, DateTime date)
        {
            EventId = eventId;
            EventName = eventName;
            Date = date;
        }

        public virtual void RegisterParticipant(string participantName)
        {
            Participants.Add(participantName);
            Console.WriteLine("Участник " + participantName + " зарегистрирован на " + EventName);
        }

        public virtual void NotifyParticipants()
        {
            Console.WriteLine("Отправка уведомления для " + Participants.Count + " участников мероприятия " + EventName);
        }

        public virtual string GetEventDetails()
        {
            return "ID: " + EventId + ", Название: " + EventName + ", Дата: " + Date.ToShortDateString();
        }
    }

    // Класс Конференция
    public class Conference : Event
    {
        public int SpeakersCount { get; set; }

        public Conference(int eventId, string eventName, DateTime date, int speakersCount)
            : base(eventId, eventName, date)
        {
            SpeakersCount = speakersCount;
        }

        public override void NotifyParticipants()
        {
            Console.WriteLine("Отправка программы конференции " + EventName + " с участием " + SpeakersCount + " спикеров.");
        }
    }

    // Класс Фестиваль
    public class Festival : Event
    {
        public int BandsCount { get; set; }

        public Festival(int eventId, string eventName, DateTime date, int bandsCount)
            : base(eventId, eventName, date)
        {
            BandsCount = bandsCount;
        }

        // Регистрация с указанием музыкального жанра
        public void RegisterParticipant(string participantName, string genre)
        {
            Participants.Add(participantName + " (" + genre + ")");
            Console.WriteLine("Участник " + participantName + " зарегистрирован на фестиваль " + EventName + " с жанром " + genre);
        }
    }

    // Класс Семинар
    public class Seminar : Event
    {
        public string Topics { get; set; }

        public Seminar(int eventId, string eventName, DateTime date, string topics)
            : base(eventId, eventName, date)
        {
            Topics = topics;
        }

        public override string GetEventDetails()
        {
            return base.GetEventDetails() + ", Темы семинара: " + Topics;
        }
    }

        Conference conf = new Conference(1, "Tech Conference", new DateTime(2026, 10, 15), 10);
        Festival fest = new Festival(2, "Rock Fest", new DateTime(2026, 7, 20), 5);
        Seminar sem = new Seminar(3, "Программирование C#", new DateTime(2026, 11, 5), "ООП, Классы, Наследование");

        Console.WriteLine("--- Регистрация ---");
        conf.RegisterParticipant("Иван");
        fest.RegisterParticipant("Анна", "Рок");
        sem.RegisterParticipant("Петр");

        Console.WriteLine("\n--- Уведомления ---");
        conf.NotifyParticipants();
        fest.NotifyParticipants();

        Console.WriteLine("\n--- Детали мероприятий ---");
    Console.WriteLine(conf.GetEventDetails());
    Console.WriteLine(fest.GetEventDetails());
     Console.WriteLine(sem.GetEventDetails());
```
